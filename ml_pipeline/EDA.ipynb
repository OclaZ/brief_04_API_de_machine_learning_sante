#Loading the Data 

```python
import pandas as pd
path = "data.csv"
df = pd.read_csv(path)
df.head()
df.info()
df.describe()
```

```python
df.isnull().sum()
```

separe les columns : catégorielles et numériques

```python
def is_numeric(df):
    return df.select_dtypes(include=['int64', 'float64'])

def is_categorie(df):
    return df.select_dtypes(exclude=['int64', 'float64'])
Data_Numeric = is_numeric(df)
Data_categorie = is_categorie(df)
Data_categorie.head()
```

Visualisation des columns 

```python
import matplotlib.pyplot as plt
import seaborn as sns

corr = Data_Numeric.corr()
plt.figure(figsize=(10, 6))
sns.heatmap(corr, vmax=1 , vmin=-1 , annot=True)
plt.title('Heatmap de corrélation des variables numériques')
plt.show()
```
